```python
import networkx as nx
import matplotlib.pyplot as plt

data = [
    ("Disease", "assoiciatewith", "LungCancer"),
    ("Disease", "assoiciatewith", "MetastaticLungCancer"),
    ("Disease", "assoiciatewith", "nonSmallCellLungCancer"),
    ("Disease", "assoiciatewith", "PulmonaryFibrosis"),
    ("Disease", "assoiciatewith", "PleuralEffusion"),
    ("PleuralEffusion", "hassympotm", "cheastpain"),
    ("PleuralEffusion", "hassympotm", "shortnessofbreath"),
    ("PleuralEffusion", "hassympotm", "cough"),
    ("PleuralEffusion", "hassympotm", "fatigue"),
    ("lungcancer", "canbetratedwith", "Chemotherapy"),
    ("lungcancer", "canbetratedwith", "surgery"),
    ("lungcancer","canbetratedwith","Targetedtheraphy"),
    ("lungcancer","disgonesewith","Biopsy"),
    ("lungcancer","disgonesewith","MRI"),
    ("lungcancer","disgonesewith","CTscan"),
    ("lungcancer","disgonesewith","petscan"),
    ("lungcancer", "hassympotm", "cough"),
    ("nonSmallCellLungCancer", "hassympotm", "cough"),

]

# Create a directed graph
G = nx.DiGraph()

# Add edges based on data
for source, relationship, target in data:
    G.add_edge(source, target, label=relationship)

# Generating a graph layout
pos = nx.spring_layout(G)

# Extract edge labels for visualization
edge_labels = nx.get_edge_attributes(G, "label")

# Drawing the graph
plt.figure(figsize=(10, 7))
nx.draw(G, pos, with_labels=True, node_color="lightblue", node_size=3000, font_size=10, font_weight="bold")
nx.draw_networkx_edge_labels(G, pos, edge_labels=edge_labels, font_size=9)
plt.title("Healthcare Knowledge Graph Visualization for lung cancer")
plt.show()
```
